## Introduction: Stochastic Gradient Descent

```python
import numpy as np

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

## Understanding the Data

```python
X,y = load_diabetes(return_X_y=True)

print(X.shape)
print(y.shape)
```

## Splitting the Data

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)

print(X_train.shape)
print(X_test.shape)
```

## Training Scikit Learn Linear Regression on the Data

```python
reg = LinearRegression()

reg.fit(X_train,y_train)

print("Intercept : ",reg.intercept_)
print("Coefficients : ",reg.coef_)
```

## Accuracy of Linear Regression

```python
y_pred = reg.predict(X_test)

print("Accuracy : ",r2_score(y_test,y_pred))
```

## Implementing Stochastic Gradient Descent

```python
class StochasticRegressor:

    def __init__(self,learning_rate=0.01,epochs=50):

        self.intercept_ = None
        self.coef_ = None
        self.lr = learning_rate
        self.ep = epochs

    def fit(self,X_train,y_train):

        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.ep):

            for j in range(X_train.shape[0]):

                idx = np.random.randint(0,X_train.shape[0])

                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_

                intercept_der = -2 * (y_train[idx] - y_hat)
                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])

                self.intercept_ = self.intercept_ - (self.lr * intercept_der)
                self.coef_ = self.coef_ - (self.lr * coef_der)

    def predict(self,X_test):

        return np.dot(X_test,self.coef_) + self.intercept_
```

## Training SGD on the Data

```python
sgd = StochasticRegressor(learning_rate=0.02,epochs=35)

sgd.fit(X_train,y_train)

print("Intercept : ",sgd.intercept_)
print("Coefficients : ",sgd.coef_)
```

## Accuracy of SGD

```python
y_pred1 = sgd.predict(X_test)

print("Accuracy : ",r2_score(y_test,y_pred1))
```
